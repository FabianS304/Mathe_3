# Übungsblatt 1

```python
import numpy as np

a2 = np.array([[-1,0],
               [1,-5]])
w2, v2 = np.linalg.eig(a2)
print('a2 EW:', w2)
print('a2 norm. EV:', v2[:,0], v2[:,1])
print(np.array([4, 1])/np.sqrt(17))

a5 = np.array([[1,0,0],
               [0,3,2],
               [0,2,0]])
w5, v5 = np.linalg.eig(a5)
print('a5 EW:', w5)
print('a5 norm. EV:', v5[:,0], v5[:,1], v5[:,2])
print(np.array([0,2,1])*(-1)/np.sqrt(5))
```
