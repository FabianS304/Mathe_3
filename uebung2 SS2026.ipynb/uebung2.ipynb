```python
import numpy as np

def onb_1(basis):
    q = np.zeros_like(basis).astype(np.float64) # Create a zero matrix with the same shape as basis with explicit type (see argumentation below)

    q[:,0] = basis[:,0] / np.linalg.norm(basis[:,0])
    for i in range(basis.shape[0]-1):
        q[:,i+1] = basis[:,i+1]
        for j in range(i+1):
            q[:,i+1] -= q[:,j] * np.dot(q[:,j], basis[:,i+1])
        q[:,i+1] /= np.linalg.norm(q[:,i+1])
    return q

def onb_2(basis):
    q, r = np.linalg.qr(basis)
    return q
```

```python
b1 = np.array([[2, 1],
              [1, 2]])

print(onb_1(b1))
print(onb_2(b1))
```
