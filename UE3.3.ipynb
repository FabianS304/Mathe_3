```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
def fit(x, y):
    X = np.hstack((np.ones_like(x), x))
    result = np.linalg.inv(X.T @ X)
    return result @ X.T @ y


def plot_fit(x, y, model):
    plt.scatter(x, y, alpha=0.5)
    plt.xlabel("x")
    plt.ylabel("y")
    t = (min(x), max(x))
    plt.plot(t, model.predict(t), "-g")
    plt.show()
```

### Anwendung 1

```python
x1 = np.array([12, 14, 17, 16, 11]).reshape(-1, 1)
y1 = np.array([55, 63, 67, 70, 51])

model = LinearRegression()
model.fit(x1, y1)
intercept = model.intercept_
slope = model.coef_[0]
print("intercept [a]:", np.around(intercept, 3))
print("slope [b]:", np.around(slope, 3))
plot_fit(x1, y1, model)
pred = model.predict(np.array([15]).reshape(-1, 1))[0]
print(f"Von einem Angestellten mit 15 Punkten im Test können etwa {np.round(pred):.0f} Einheiten erwartet werden.")

print("[a,b] =", fit(x1,y1))
```

### Anwendung 2

```python
x2 = np.array([161, 169, 176, 185, 189]).reshape(-1, 1)
y2 = np.array([52, 59, 71, 80, 83])

model = LinearRegression()
model.fit(x2, y2)
intercept = model.intercept_
slope = model.coef_[0]
print("intercept [a]:", np.around(intercept, 3))
print("slope [b]:", np.around(slope, 3))
plot_fit(x2, y2, model)
pred = model.predict(np.array([182]).reshape(-1, 1))[0]
print(f"Ein 182 cm großer Mensch bringt etwa {np.round(pred):.0f} kg auf die Waage.")

print("[a,b] =", np.around(fit(x2,y2),3))
```
